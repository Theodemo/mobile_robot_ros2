```python
import numpy
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
xdata = numpy.array([0.01, 25, 50, 75, 100, 125, 150, 175, 200, 225, 250], dtype=numpy.float128)
ydata = numpy.array([700, 480, 350, 280, 235, 200, 180, 160, 140, 130, 110], dtype=numpy.float128)

#fx_weights = numpy.polyfit(x_pt, y_pt, 3)
def ir_fx(x, a):
    return 590 * numpy.exp(-a * x) + 110

popt, pcov = curve_fit(ir_fx, xdata, ydata, maxfev=100000)

print(popt)

plt.plot(xdata, ydata, marker='x',label = "points")
plt.plot(xdata, ir_fx(xdata, *popt), marker='x',label = "estimate")
plt.legend()
```

```python
plt.plot(xdata, ydata, marker='x',label = "points")
plt.plot(xdata, ir_fx(xdata,0.0164), marker='x',label = "points")
plt.legend()

#ydata = 590*exp(-0.0164*x) +110
# distance(mm) = 590*exp(-0.0164*AD_Output) +110
```

```python
from matplotlib.backends.backend_qt5agg import FigureCanvasQTAgg as FigureCanvas
from matplotlib.backends.backend_qt5agg import FigureCanvasQTAgg, NavigationToolbar2QT as NavigationToolbar
from matplotlib.figure import Figure
```

```python
class MplCanvas(FigureCanvas):

    def __init__(self, parent=None, width=1, height=2, dpi=100):
        fig = Figure(figsize=(width, height), dpi=dpi)
        self.motor_axes = fig.add_subplot(211) 
        self.robot_axes = fig.add_subplot(212)
        super(MplCanvas, self).__init__(fig)
```

```python
canvas = MplCanvas(width=5, height=4, dpi=100)
```
